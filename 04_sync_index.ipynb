```python
from databricks.ai_search.client import AISearchClient

client = AISearchClient(disable_notice=True)
index = client.get_index(
    endpoint_name="privacylens-endpoint",
    index_name="workspace.privacylens.silver_chunks_index",
)
index.sync()
print("Index sync triggered")
```

```python
%sql
SELECT citation, search_score
FROM vector_search(
  index => 'workspace.privacylens.silver_chunks_index',
  query_text => 'Can a user make a company delete their data?',
  num_results => 3
);
```
